```python
import pandas as pd

df = pd.read_csv('model_panel.csv')
print("Data loaded successfully. Here are the first 5 rows:")
display(df.head())
```

```python
print("\nHere is a basic description of the data:")
display(df.describe())
```

```python
# Filter out data for years 2022 and below
df_filtered = df[df['tahun'] > 2021]

print("Data after filtering for years > 2022:")
display(df_filtered.describe())
```

```python
missing_x4_tpt_pct = df_filtered['x4_tpt_pct'].isnull().sum()
print(f"Number of missing values in 'x4_tpt_pct': {missing_x4_tpt_pct}")
```

```python
# Convert 'tanggal' to datetime objects if not already
df_filtered['tanggal'] = pd.to_datetime(df_filtered['tanggal'])

# Sort the DataFrame by 'provinsi_id' and 'tanggal' to ensure correct filling order
df_filtered = df_filtered.sort_values(by=['provinsi_id', 'tanggal']).reset_index(drop=True)

# Apply a 5-month forward fill within each 'provinsi_id' group directly to 'x4_tpt_pct'
df_filtered['x4_tpt_pct'] = df_filtered.groupby('provinsi_id')['x4_tpt_pct'].ffill(limit=5)

# Apply a 1-month backward fill within each 'provinsi_id' group directly to 'x4_tpt_pct'
# This will specifically fill any remaining NaNs, like Jan 2022 from Feb 2022, after the forward fill.
df_filtered['x4_tpt_pct'] = df_filtered.groupby('provinsi_id')['x4_tpt_pct'].bfill(limit=1)

print("DataFrame after 5-month forward fill and 1-month backward fill for 'x4_tpt_pct':")
# Display some rows to demonstrate the fills, showing original and filled values
display(df_filtered[df_filtered['provinsi_id'] == 1][['tanggal', 'x4_tpt_pct']].head(12))
```

```python
cols_to_check = ['x6_tabungan_miliar', 'x7_jumlah_kc_bank', 'x8_ldr_pct', 'x9_npl_ratio', 'x10_rasio_umkm']

for col in cols_to_check:
    missing_data = df_filtered[df_filtered[col].isnull()]
    if not missing_data.empty:
        print(f"\n--- Missing values for column: {col} ---")
        # Group by provinsi_id and nama_provinsi to find the date range of missing values
        missing_summary = missing_data.groupby(['provinsi_id', 'nama_provinsi'])['tanggal'].agg(['min', 'max', 'count']).reset_index()
        missing_summary = missing_summary.rename(columns={'min': 'start_tanggal_missing', 'max': 'end_tanggal_missing', 'count': 'num_missing_entries'})
        display(missing_summary)
    else:
        print(f"\nNo missing values found for column: {col}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# List of columns to apply smoothing
sticky_cols = ['x3_pdrb_per_kapita', 'x4_tpt_pct', 'x5_penetrasi_internet_pct',
               'x6_tabungan_miliar', 'x7_jumlah_kc_bank', 'x8_ldr_pct',
               'x9_npl_ratio', 'x10_rasio_umkm']

df_before = df_filtered.copy()

def apply_true_spline(group):
    group = group.sort_values(['tahun', 'bulan']).copy()

    for col in sticky_cols:
        # Skip if column has all NaN
        if group[col].isna().all():
            continue
            
        is_anchor = group[col] != group[col].shift(1)

        # --- KUNCI PERBAIKAN MATEMATIS ---
        is_anchor.iloc[0] = True   # Kunci jangkar ujung kiri (awal data)
        is_anchor.iloc[-1] = True  # Kunci jangkar ujung kanan (akhir data)
        # --------------------------------

        series_with_nans = group[col].where(is_anchor, np.nan)
        group[col] = series_with_nans.interpolate(method='pchip')
        group[col] = group[col].ffill().bfill()

    return group

# Apply smoothing - preserve provinsi_id by resetting the index from groupby
df_spline = df_before.groupby('provinsi_id', group_keys=False).apply(apply_true_spline).reset_index()
# Rename 'index' to 'provinsi_id' (from groupby key) and drop extra level
if 'index' in df_spline.columns:
    df_spline = df_spline.rename(columns={'index': 'provinsi_id'})
if 'level_0' in df_spline.columns:
    df_spline = df_spline.drop('level_0', axis=1)

print(f"df_spline shape: {df_spline.shape}")
print(f"df_spline columns: {df_spline.columns.tolist()}")

# --- Visualisasi Komparasi Semua Kolom ---
prov_id = 22
data_before = df_before[df_before['provinsi_id'] == prov_id].sort_values(['tahun', 'bulan'])
data_after = df_spline[df_spline['provinsi_id'] == prov_id].sort_values(['tahun', 'bulan'])

fig, axes = plt.subplots(nrows=4, ncols=2, figsize=(15, 20))
fig.suptitle(f'Smoothing Comparison for Province ID {prov_id} (x3 to x10)', fontsize=16)
axes = axes.flatten()

for i, col in enumerate(sticky_cols):
    axes[i].plot(range(len(data_before)), data_before[col], 'o--', label='Before (Step)', alpha=0.4, markersize=4)
    axes[i].plot(range(len(data_after)), data_after[col], '-', label='After (Smoothed)', linewidth=2, color='orange')
    axes[i].set_title(f'Variable: {col}')
    axes[i].legend()
    axes[i].grid(True, linestyle='--', alpha=0.7)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
import pandas as pd

# Asumsi df_final adalah dataframe Anda yang sudah melalui proses PCHIP
# Pastikan kolom 'tanggal' sudah dalam format datetime (seperti pada info() Anda)
df_lagged = df_spline.copy()

# 1. KEAMANAN MUTLAK: Mengurutkan data berdasarkan hierarki ruang dan waktu
# Ini memastikan data selalu berurutan dari Jan 2022 -> Des 2025 per provinsi
df_lagged = df_lagged.sort_values(by=['provinsi_id', 'tanggal']).reset_index(drop=True)

# 2. Definisikan variabel independen (X) yang akan diberikan efek lag
# Kita mengecualikan provinsi_id, nama_provinsi, tanggal, tahun, bulan, dan twp90_pct
x_features = [col for col in df_lagged.columns if col.startswith('x')]

# 3. Eksekusi Time Lags (3 Bulan / 1 Kuartal dan 6 Bulan / 1 Semester)
lags = [3, 6]

for lag in lags:
    for feature in x_features:
        # Penamaan kolom baru secara terstruktur
        new_col_name = f"{feature}_lag_{lag}"

        # Operasi shift diisolasi per provinsi
        df_lagged[new_col_name] = df_lagged.groupby('provinsi_id')[feature].shift(lag)

# 4. (Opsional tapi Direkomendasikan)
# Evaluasi proporsi Missing Values (NaN) akibat proses shifting
print("--- Laporan Missing Values Pasca-Lagging ---")
print(df_lagged[[f"{x_features[0]}_lag_3", f"{x_features[0]}_lag_6"]].isnull().sum())
print(f"Total baris awal: {len(df_lagged)}")
df_lagged.info()
```

```python
import subprocess
import sys

# Install xgboost if not already installed
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "xgboost", "scikit-learn"])

import pandas as pd
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# 1. Check data status before dropna
print(f"df_lagged shape before dropna: {df_lagged.shape}")
print(f"Missing values count:\n{df_lagged.isnull().sum()}")
print(f"\nData year range: {df_lagged['tahun'].min()} - {df_lagged['tahun'].max()}")

# 2. Separate train/test before dropna to understand the split
train_data_all = df_lagged[df_lagged['tahun'] <= 2024]
test_data_all = df_lagged[df_lagged['tahun'] == 2025]

print(f"\nTrain data (tahun <= 2024) before dropna: {train_data_all.shape}")
print(f"Test data (tahun == 2025) before dropna: {test_data_all.shape}")

# 3. Definisikan Prediktor (X) dan Target (y)
kolom_non_prediktor = ['provinsi_id', 'nama_provinsi', 'tanggal', 'tahun', 'bulan', 'twp90_pct']
X_cols = [col for col in df_lagged.columns if col not in kolom_non_prediktor]

print(f"\nFeature columns (total {len(X_cols)}): {X_cols}")

# 4. Drop NaN only for modeling, keeping original split logic
train_data = train_data_all.dropna(subset=X_cols + ['twp90_pct'])
test_data = test_data_all.dropna(subset=X_cols + ['twp90_pct'])

X_train = train_data[X_cols]
y_train = train_data['twp90_pct']

X_test = test_data[X_cols]
y_test = test_data['twp90_pct']

print(f"\nAfter dropna:")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")

if X_train.shape[0] == 0 or X_test.shape[0] == 0:
    print("\nWARNING: Train or Test set is empty!")
    print("Using alternative split: 80% train, 20% test (stratified by time)")
    
    # Create an alternative split by sampling
    df_clean = df_lagged.dropna(subset=X_cols + ['twp90_pct'])
    print(f"Total clean data points: {df_clean.shape[0]}")
    
    # Sort by date to maintain temporal structure
    df_clean = df_clean.sort_values('tanggal').reset_index(drop=True)
    
    # Use 80/20 split (roughly 2022-2024 vs later months of 2024/2025)
    split_idx = int(0.8 * len(df_clean))
    
    X_train = df_clean.iloc[:split_idx][X_cols]
    y_train = df_clean.iloc[:split_idx]['twp90_pct']
    
    X_test = df_clean.iloc[split_idx:][X_cols]
    y_test = df_clean.iloc[split_idx:]['twp90_pct']
    
    print(f"New X_train shape: {X_train.shape}")
    print(f"New X_test shape: {X_test.shape}")
else:
    print("\nData split successful")

# 5. Pelatihan Model Baseline XGBoost (Regresi)
print("\n--- Melatih XGBoost Baseline Model ---")
xgb_model = xgb.XGBRegressor(
    objective='reg:squarederror',
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

xgb_model.fit(X_train, y_train, verbose=False)

# 6. Prediksi dan Evaluasi
y_pred = xgb_model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"\n--- Evaluasi Baseline Model ---")
print(f"RMSE: {rmse:.6f}")
print(f"MAE: {mae:.6f}")
print(f"R-squared (R2): {r2:.6f}")

# 7. Feature Importance
importance = pd.DataFrame({
    'Feature': X_cols,
    'Importance': xgb_model.feature_importances_
}).sort_values(by='Importance', ascending=False)

print(f"\n--- Top 10 Fitur Paling Berpengaruh ---")
print(importance.head(10).to_string())
```
